```python
import pandas as pd
import matplotlib.pyplot as plt
import argparse
import os
import sys

def visualize(participant_folder):

    airflow = pd.read_csv(f"{participant_folder}/nasal_airflow.txt", names=["time","value"])
    thoracic = pd.read_csv(f"{participant_folder}/thoracic_movement.txt", names=["time","value"])
    spo2 = pd.read_csv(f"{participant_folder}/spo2.txt", names=["time","value"])

    events = pd.read_csv(f"{participant_folder}/flow_events.csv")

    fig, axs = plt.subplots(3,1, figsize=(15,8), sharex=True)

    axs[0].plot(airflow["time"], airflow["value"])
    axs[0].set_title("Nasal Airflow")

    axs[1].plot(thoracic["time"], thoracic["value"])
    axs[1].set_title("Thoracic Movement")

    axs[2].plot(spo2["time"], spo2["value"])
    axs[2].set_title("SpO2")

    for _, row in events.iterrows():
        start = row["start_time"]
        end = row["end_time"]

        for ax in axs:
            ax.axvspan(start, end, color='red', alpha=0.3)

    os.makedirs("Visualizations", exist_ok=True)

    name = os.path.basename(participant_folder)

    plt.savefig(f"Visualizations/{name}_visualization.pdf")

    plt.close()

if __name__ == "__main__":


    participant_folder = "path/to/your/participant_folder"

    visualize(participant_folder)
```
